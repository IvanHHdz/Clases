```python
from math import lcm, gcd
```

```python
p = 101
q = 113
m = lcm(p - 1, q - 1)
n = p*q
e = 3
# gcd(5, m) # = 1
d = ((m + e - 1) // e * (e - 1)) - e + 2
# (d * e) % m
print(f"Público: ({n}, {e})")
print(f"Privado: d = {d}")
```

```python
def cifrar(str):
    # arreglo para meter las letras individualmente
    letras = []
    # aqui guardamos letra por letra
    for c in str:
        # convertimos antes cada letra a decimal
        letras.append(ord(c))
    print(letras)
    # arreglo para los datos cifrados
    datos = []
    for l in letras:
        k = l**e % n
        datos.append(k)
    return datos
```

```python
def descifrar(datos):
    # arreglo para guardar las letras descifradas
    letras = []
    for k in datos:
        # aplicamos la clave privada y recuperamos el código de la letra
        codigo = pow(k, d, n)
        letras.append(chr(codigo))
    return "".join(letras)
```
